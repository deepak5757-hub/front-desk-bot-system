```python
 user = {"Unique Patient ID " :[],
            "Patient Name" :[],
            "Age ":[],
            "Date of Birth":[],
            "Gender":[],
            "Mobile Number" : [],
            "Address":[]}
```

```python
import datetime as dt
```

```python
def user_info():
    while True: # while true ye karta h ki jab jab condition fals hogi tab tab qution puchta jayega
        name = input('please enter patient name with following MR.| Mrs. | or |MISS. ?')
        if name [: name.find('.')].lower() in ['mr','mrs','miss'] and name [name.find('.')+1:].replace(" ","").isalpha():
            user["Patient Name"].append(name)
            break
        else: # agar condition k according na hoga to else calega
            print("please enter correct name for eg: MR.john dayne")
    while True:
        DOB = input("please enter patient date of birth in YYYY-MM-DD")
        try:
            DOB = dt.datetime.strptime(DOB,"%Y-%m-%d")
            if DOB<=dt.datetime.now(): # ye check karega ki courent date and time se jayada na hona chiya
                user["Date of Birth"].append(DOB)
                break
        except:
            print("please enter date in valid format YYYY-MM-DD")
    while True:
        Gender = input('please enter Gender M for male and F for female ?')
        if Gender.lower() in ['m' , 'f']:
            user['Gender'].append(Gender)
            break
        else:
            print('please enter Gender as M OR F ')
    while True:
        Number = input('pkease enter your Mobile number followed by +91 ?')
        if Number[:4] =='+91-' and len(Number)==14 : # 14 is leya q ki +91- ye bhii to 4 h
            user['Mobile Number'].append(Number)
            break
        else:
            print('please enter number with +91-99999999999 format')
    while True:
        address = input('please enter patient address ')
        if address == '':
            user['Address'].append('delhi')
            break
        elif len(address)<=100:
            user['Address'].append(address)
            break
        else:
            print('please enter Address in less than 100 characters')
    if user['Unique Patient ID '] ==[]:
        user['Unique Patient ID '].append(1)
    else:
        user["Unique Patient ID "].append(max(user['Unique Patient ID '])+1)
        user['Age'].append(int(abs((DOB-dt.datetime.now()).days/365)))
        
```
